```python
# obtenemos un cursor que utilizamos para las queries
crsr = connection.cursor()
```

```python
def sql_query(query):

    # Ejecuta la query
    crsr.execute(query)

    # Almacena los datos de la query 
    ans = crsr.fetchall()

    # Obtenemos los nombres de las columnas de la tabla
    names = [description[0] for description in crsr.description]

    return pd.DataFrame(ans,columns=names)
```

```python
import pandas as pd
import glob

# Ruta donde están tus CSV
ruta_csv = "ruta/a/tus/csv/*.csv"

# Leer todos los CSV y quedarnos solo con las columnas necesarias
dfs = []
for archivo in glob.glob(ruta_csv):
    df = pd.read_csv(archivo, usecols=["nombre", "email", "promocion"])
    dfs.append(df)

# Unir todos los DataFrames por filas
df_final = pd.concat(dfs, ignore_index=True)

# Crear la columna primary key numérica consecutiva como string
df_final.insert(0, "id", df_final.index.astype(str))

# Guardar el resultado en un nuevo CSV listo para importar a SQL
df_final.to_csv("resultado_final.csv", index=False)

print("Archivo combinado creado: resultado_final.csv")
'''Qué hace este script:

Lee todos los CSV de la carpeta indicada.
Selecciona solo las columnas nombre, email y promocion.
Combina las filas de todos los archivos.
Inserta una columna id al principio con valores consecutivos ("0", "1", "2", …) en formato string.
Guarda el resultado en un nuevo CSV.
💡 Si quieres que la primary key empiece en "1" y no en "0", puedes cambiar:

Python

Copiar código'''
df_final.insert(0, "id", (df_final.index + 1).astype(str))
```
