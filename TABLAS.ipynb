TABLA ALUMNOS

```python
df1 = pd.read_csv(archivo1, sep=";")[["Nombre", "Email"]]
df2 = pd.read_csv(archivo2, sep=";")[["Nombre", "Email"]]
df3 = pd.read_csv(archivo3, sep=";")[["Nombre", "Email"]]
df4 = pd.read_csv(archivo4, sep=";")[["Nombre", "Email"]]
```

```python
df_alumnos = pd.concat([df1, df2, df3, df4], ignore_index=True)
df_alumnos.insert(0, "id_alumno", range(1, len(df_alumnos) + 1))
df_alumnos.rename(columns={"Nombre": "nombre_alumno", "Email": "email"}, inplace=True)
```

```python
conn = sqlite3.connect("alumnos.db")
cursor = conn.cursor()
```

```python
conn.execute("""
CREATE TABLE alumnos (
    "id_alumno" INTEGER PRIMARY KEY,
    nombre_alumno VARCHAR(100) NOT NULL,
    email VARCHAR(150)
)
""")

df_alumnos.to_sql(
    "alumnos",
    conn,
    if_exists="append",
    index=False
)

df_alumnos.to_sql(
    "alumnos",
    conn,
    if_exists="replace",
    index=False
)
conn.commit()
conn.close()
```
